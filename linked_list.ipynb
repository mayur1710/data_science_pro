**1. Reverse a Linked List In-Place**

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverse_linked_list(head):
    prev = None
    current = head
    while current:
        next_node = current.next  # Save the next node
        current.next = prev  # Reverse the link
        prev = current  # Move prev to current
        current = next_node  # Move to next node
    return prev  # New head of the reversed list
```

Explanation:

Initialization: Start with prev as None and current pointing to the head of the list.

Iterate: For each node, reverse the link by pointing current.next to prev.

Move Forward: Update prev to be the current node and current to be next_node.

Termination: When the loop ends, prev will be the new head of the reversed list.

**2. Detect Cycle in a Linked List**

```python
def has_cycle(head):
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next  # move slow pointer by one step
        fast = fast.next.next  # move fast pointer by two steps
        if slow == fast:
            return True  # cycle detected
    return False  # no cycle
```

Explanation:

Two Pointers: slow moves one step at a time, while fast moves two steps at a time.

Cycle Detection: If there is a cycle, slow and fast will eventually meet. If fast reaches the end, there is no cycle.

**3. Merge Two Sorted Linked Lists**

```python
def merge_two_lists(l1, l2):
    dummy = ListNode()
    current = dummy
    while l1 and l2:
        if l1.val < l2.val:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next
    current.next = l1 if l1 else l2  # Attach the remaining part
    return dummy.next  # The merged list starts from dummy.next
```

Explanation:

Dummy Node: Use a dummy node to simplify the merging process.

Iteration: Compare the values from l1 and l2, append the smaller one to the merged list, and move the pointer forward.

Remaining Nodes: Once one list is exhausted, attach the remaining nodes from the other list.

**4. Remove N-th Node from the End of the List**

```python
def remove_nth_from_end(head, n):
    dummy = ListNode(0, head)
    fast = slow = dummy
    for _ in range(n):
        fast = fast.next  # move fast n steps ahead
    while fast.next:
        fast = fast.next
        slow = slow.next
    slow.next = slow.next.next  # remove the nth node
    return dummy.next
```

Explanation:

Two Pointers: fast moves n steps ahead of slow. Then, both move together until fast reaches the end.

Remove Node: slow will be just before the node that needs to be removed.

**5. Remove Duplicates from a Sorted Linked List**

```python
def remove_duplicates(head):
    current = head
    while current and current.next:
        if current.val == current.next.val:
            current.next = current.next.next  # skip the duplicate
        else:
            current = current.next
    return head
```

Explanation:
Check Duplicates: If the current node and the next node have the same value, bypass the next node.

Move Forward: Continue this process until all duplicates are removed.

**6. Find the Intersection of Two Linked Lists**

```python
def get_intersection_node(headA, headB):
    if not headA or not headB:
        return None
    a, b = headA, headB
    while a != b:
        a = a.next if a else headB
        b = b.next if b else headA
    return a
```

Explanation:
Two Pointers: a and b start at the heads of the two lists. When they reach the end, they switch to the other list.

Intersection: If the lists intersect, a and b will meet at the intersection node; otherwise, they will both reach None.

**7. Rotate a Linked List by K Positions to the Right**

```python
def rotate_right(head, k):
    if not head or not head.next or k == 0:
        return head
    # Compute the length of the list
    length = 1
    old_tail = head
    while old_tail.next:
        old_tail = old_tail.next
        length += 1
    # Connect the tail to head to make it circular
    old_tail.next = head
    # Find new tail and new head
    new_tail = head
    for _ in range(length - k % length - 1):
        new_tail = new_tail.next
    new_head = new_tail.next
    new_tail.next = None  # Break the loop
    return new_head
```

Explanation:

Circular List: Connect the tail to the head to make the list circular.

New Head: Calculate the position of the new head after rotation by moving length - k % length steps from the head.

**8. Add Two Numbers Represented by Linked Lists**

```python
def add_two_numbers(l1, l2):
    dummy = ListNode()
    current = dummy
    carry = 0
    while l1 or l2 or carry:
        val1 = l1.val if l1 else 0
        val2 = l2.val if l2 else 0
        total = val1 + val2 + carry
        carry = total // 10
        current.next = ListNode(total % 10)
        current = current.next
        if l1:
            l1 = l1.next
        if l2:
            l2 = l2.next
    return dummy.next
```

**9. Clone a Linked List with Next and Random Pointer**

Solution:
This can be done in three steps:

Create a copy of each node and link them side-by-side in the original list.
Assign the random pointers to the copied nodes.
Separate the copied list from the original list.

```python
class Node:
    def __init__(self, val, next=None, random=None):
        self.val = val
        self.next = next
        self.random = random

def copy_random_list(head):
    if not head:
        return None
    # Step 1: Create a copy of each node
    current = head
    while current:
        new_node = Node(current.val, current.next)
        current.next = new_node
        current = new_node.next
    # Step 2: Assign random pointers to the copied nodes
    current = head
    while current:
        current.next.random = current.random.next if current.random else None
        current = current.next.next
    # Step 3: Separate the copied list from the original list
    old_list = head
    new_list = head.next
    new_head = head.next
    while old_list:
        old_list.next = old_list.next.next
        new_list.next = new_list.next.next if new_list.next else None
        old_list = old_list.next
        new_list = new_list.next
    return new_head
```

Explanation:
Copy Nodes: First, copy each node and link it to the original node's next.

Assign Random Pointers: Then, correctly assign the random pointers.

Separate Lists: Finally, restore the original list and separate the copied list.
